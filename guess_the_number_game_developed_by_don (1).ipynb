```python
import random

def guessgame():
    print("welcome to number guessing game")
    print("enter a number between 1to 100")
    try:
        secret_num = random.randint(1, 100)
        for i in range(1, 8):
            guess = int(input("Enter a number: "))
            if guess < 1 or guess > 100:
                print("Invalid number. Please enter a number between 1 and 100.")
                continue
            elif guess == secret_num:
                print(f"Congratulations! You guessed the number {guess} in {i} attempts.")
                break
            elif guess < secret_num:
                print("Your number is greater than the secret number.")
            else:
                print("Your number is lower than the secret number.")
        else:
            print("Your limit exceeds! You have used all 7 attempts.")
            print(f"The secret number was {secret_num}.")
    except ValueError:
        print("Invalid only numbers are allowed")

guessgame()
```
